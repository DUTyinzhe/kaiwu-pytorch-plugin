# Halide Perovskite — 能量泛函 QUBO + CIM
## 图半监督学习: 材料相似图 → 图拉普拉斯 → $\min \frac{1}{2}x^T A x - x^T b$ → CIM
36维元素描述符 → k-NN图 → 少数已知标签 → 能量泛函 QUBO → CIM 求解全节点带隙

```python
import numpy as np
import pandas as pd
import kaiwu as kw
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import StandardScaler
import warnings, json, os
warnings.filterwarnings('ignore')
kw.common.CheckpointManager.save_dir = '/tmp'

BIT_WIDTH = 8
K_NN = 8
LAMBDA_LABEL = 10.0        # 标记节点惩罚系数
EPS_REG = 0.01             # 图拉普拉斯正则化 (保证 SPD)
N_SUBGRAPH = 60            # 子图节点数
N_LABELED = 10             # 已知标签数
BOUND_MARGIN = 0.5

OUTPUT_DIR = 'D:/QPDE/photo+kan/outputs'
os.makedirs(OUTPUT_DIR, exist_ok=True)
print('模块加载完成')
```

```python
# ===== 加载 Halide Perovskite 数据 =====
# CSV 无表头, 58列: 0=formula, 1=mixing, 2-7=PBE属性, 8-21=组分描述符, 22-57=元素描述符(36列)
url = 'https://raw.githubusercontent.com/mannodiarun/halide_perovs_design/main/PBE_data.csv'
df = pd.read_csv(url, header=None)
target_raw = df.iloc[:, 3].values.astype(float)  # PBE_gap
features_raw = df.iloc[:, 22:58].values.astype(float)  # 36 elemental descriptors

print(f'全部: {df.shape[0]} 条, 特征: {features_raw.shape[1]} 维')
print(f'PBE_gap 范围: [{target_raw.min():.4f}, {target_raw.max():.4f}]')
```

```python
# ===== 特征标准化 + 采样子图 =====
scaler = StandardScaler().fit(features_raw)
features = scaler.transform(features_raw)

rng = np.random.default_rng(42)
idx_sample = rng.choice(len(features), N_SUBGRAPH, replace=False)
feat_sub = features[idx_sample]
target_sub = target_raw[idx_sample]
print(f'子图: {N_SUBGRAPH} 个材料')
```

```python
# ===== 构建 k-NN 相似图 + 对称归一化拉普拉斯 =====
sim = cosine_similarity(feat_sub)
np.fill_diagonal(sim, -np.inf)
W = np.zeros((N_SUBGRAPH, N_SUBGRAPH))
for i in range(N_SUBGRAPH):
    top_k = np.argpartition(-sim[i], K_NN)[:K_NN]
    W[i, top_k] = sim[i, top_k]
W = (W + W.T) / 2  # 对称化

D_vec = W.sum(axis=1)
D_inv_sqrt = np.diag(1.0 / np.sqrt(np.maximum(D_vec, 1e-12)))
L_sym = np.eye(N_SUBGRAPH) - D_inv_sqrt @ W @ D_inv_sqrt
L_sym = (L_sym + L_sym.T) / 2
print(f'图: 边数={(W>0).sum()}, 度范围=[{D_vec.min():.1f},{D_vec.max():.1f}]')
```

```python
# ===== 标记节点 & 能量泛函系统 A x = b =====
# E(x) = (1/2) x^T L_sym x + (λ/2) Σ_{i∈L} (x_i - y_i)^2
#      = (1/2) x^T (L_sym + λ I_L + ε I) x - x^T (λ I_L y)
labeled_idx = rng.choice(N_SUBGRAPH, N_LABELED, replace=False)
labeled_mask = np.zeros(N_SUBGRAPH, dtype=bool)
labeled_mask[labeled_idx] = True

I_L = np.diag(labeled_mask.astype(float))
A_mat = L_sym + LAMBDA_LABEL * I_L + EPS_REG * np.eye(N_SUBGRAPH)
b_vec = LAMBDA_LABEL * labeled_mask.astype(float) * target_sub

print(f'A = L_sym + λ I_L + ε I: {A_mat.shape}, κ(A)={np.linalg.cond(A_mat):.2e}')
print(f'标记: {N_LABELED} 个, 未知: {N_SUBGRAPH - N_LABELED} 个')
```

```python
# ===== 经典参考解 & 变量边界 =====
x_ref = np.linalg.solve(A_mat, b_vec)
margin = BOUND_MARGIN
rng_val = max(x_ref.max() - x_ref.min(), 0.1)
lb = x_ref.min() - margin * rng_val
ub = x_ref.max() + margin * rng_val
print(f'参考解范围: [{x_ref.min():.4f}, {x_ref.max():.4f}]')
print(f'编码界: [{lb:.4f}, {ub:.4f}]')

lb_vec = np.full(N_SUBGRAPH, lb)
ub_vec = np.full(N_SUBGRAPH, ub)
print(f'QUBO 变量: {N_SUBGRAPH} (全节点), 量子比特: {N_SUBGRAPH * BIT_WIDTH}')
```

```python
# ===== Binary Encoding =====
def build_binary_encoding(n_vars, bit_width, lower, upper):
    n_binary = n_vars * bit_width
    E = np.zeros((n_vars, n_binary), dtype=float)
    c = lower.copy().astype(float)
    denom = (2**bit_width) - 1.0
    for i in range(n_vars):
        step = (upper[i] - lower[i]) / denom
        for j in range(bit_width):
            E[i, i*bit_width + j] = step * (2**j)
    return E, c

def decode_binary_solution(z_binary, E, c):
    return E @ z_binary + c

E_mat, c_vec = build_binary_encoding(N_SUBGRAPH, BIT_WIDTH, lb_vec, ub_vec)
print(f'E: {E_mat.shape}, c: {c_vec.shape}, 二进制变量: {N_SUBGRAPH * BIT_WIDTH}')
```

```python
# ===== 构建能量泛函 QUBO =====
# min (1/2) x^T A x - x^T b,  x = E z + c
# = (1/2) z^T E^T A E z + z^T E^T (A c - b) + const
# Q = (1/2) E^T A E + diag(E^T (A c - b))

def build_energy_qubo(A, b, E, c):
    ETAE = E.T @ A @ E
    ETAE = (ETAE + ETAE.T) / 2  # 对称化
    w = A @ c - b
    Q = 0.5 * ETAE + np.diag(E.T @ w)
    return Q.astype(np.float32)

Q_float = build_energy_qubo(A_mat, b_vec, E_mat, c_vec)
print(f'原始 QUBO: {Q_float.shape}, 值: [{Q_float.min():.4f}, {Q_float.max():.4f}]')

Q_qubo = kw.qubo.adjust_qubo_matrix_precision(Q_float)
print(f'调整后 QUBO: [{Q_qubo.min():.4f}, {Q_qubo.max():.4f}]')
```

```python
# ===== 提交 CIM 任务 =====
ising_mat, ising_bias = kw.conversion.qubo_matrix_to_ising_matrix(Q_qubo)
n_vars = ising_mat.shape[0]
variables = [f'x[{i}]' for i in range(n_vars)]
ising_model = kw.ising.IsingModel(variables=variables, ising_matrix=ising_mat, bias=ising_bias)

print(f'Ising: {ising_mat.shape}, 变量数: {n_vars}')

optimizer = kw.cim.CIMOptimizer(task_name='perov_e3', task_mode='quota')
optimizer.solve(ising_model.get_matrix())  # solve#1: 提交
print('CIM 任务已提交 (perov_energy)，等完成后跑下一个 cell')
```

```python
# ===== 取回 CIM 结果 + 解码 =====
sol = optimizer.solve(ising_model.get_matrix())  # solve#2: 取回
print(f'CIM 返回: {sol.shape}')

solutions = sol[:, :-1]
deltas = sol[:, -1]
solutions_binary = (solutions * deltas[:, np.newaxis] + 1) / 2

energies = np.array([s @ Q_qubo @ s for s in solutions_binary])
best_idx = np.argmin(energies)
z_best = solutions_binary[best_idx]

x_quantum = decode_binary_solution(z_best, E_mat, c_vec)

print(f'最佳解索引: {best_idx}, 能量: {energies[best_idx]:.6f}')
print(f'量子解: [{x_quantum.min():.4f}, {x_quantum.max():.4f}]')
print(f'参考解: [{x_ref.min():.4f}, {x_ref.max():.4f}]')

rmse_ref = np.sqrt(np.mean((x_quantum - x_ref)**2))
rmse_dft = np.sqrt(np.mean((x_quantum - target_sub)**2))
print(f'RMSE vs 参考: {rmse_ref:.6e}, RMSE vs DFT: {rmse_dft:.6e}')
```

```python
# ===== 保存预设 =====
np.save(f'{OUTPUT_DIR}/perov_nodes_energy.npy', idx_sample)
np.save(f'{OUTPUT_DIR}/perov_features_energy.npy', feat_sub)
np.save(f'{OUTPUT_DIR}/perov_values_energy.npy', x_quantum)
np.save(f'{OUTPUT_DIR}/perov_target_energy.npy', target_sub)
np.save(f'{OUTPUT_DIR}/perov_labeled_mask_energy.npy', labeled_mask)
np.save(f'{OUTPUT_DIR}/perov_W_energy.npy', W)
np.save(f'{OUTPUT_DIR}/perov_A_energy.npy', A_mat)
np.save(f'{OUTPUT_DIR}/perov_b_energy.npy', b_vec)

meta = {
    'method': 'energy_functional_graph',
    'n_materials': int(N_SUBGRAPH),
    'n_labeled': int(N_LABELED),
    'bit_width': BIT_WIDTH,
    'qubo_size': int(n_vars),
    'best_energy': float(energies[best_idx]),
    'rmse_vs_ref': float(rmse_ref),
    'rmse_vs_dft': float(rmse_dft),
    'kappa_A': float(np.linalg.cond(A_mat))
}
with open(f'{OUTPUT_DIR}/meta_energy.json', 'w') as f:
    json.dump(meta, f, indent=2)
print(f'预设已保存, RMSE vs 参考={rmse_ref:.6e}, RMSE vs DFT={rmse_dft:.6e}')
```
